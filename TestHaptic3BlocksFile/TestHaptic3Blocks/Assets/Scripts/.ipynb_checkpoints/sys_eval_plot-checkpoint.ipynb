```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

def load_and_process_data(filepath):
    """Load and process the system performance CSV data."""
    # Read CSV and convert numeric columns
    df = pd.read_csv(filepath)
    
    # Convert all numeric columns to float
    numeric_columns = ['Timestamp', 'Connected', 'TimeSinceLastReceived', 
                      'TimeSinceLastSent', 'Port', 'MessageRate', 
                      'AverageLatency', 'MessagesSent', 'MessagesReceived', 
                      'PacketLoss', 'Stability']
    
    for col in numeric_columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    
    # Normalize timestamp to start at 0
    df['Timestamp'] = df['Timestamp'] - df['Timestamp'].min()
    return df

def create_performance_dashboard(df, output_dir='plots'):
    """Create a comprehensive performance dashboard with multiple plots."""
    # Create output directory if it doesn't exist
    Path(output_dir).mkdir(parents=True, exist_ok=True)
    
    # Set style
    plt.style.use('bmh')  # Using 'bmh' style instead of seaborn
    
    # Create a 3x2 subplot figure
    fig, axes = plt.subplots(3, 2, figsize=(15, 18))
    fig.suptitle('System Performance Dashboard', fontsize=16, y=0.95)
    
    # 1. Message Rate and Latency
    ax1 = axes[0, 0]
    ax1_twin = ax1.twinx()
    
    ax1.plot(df['Timestamp'], df['MessageRate'], 'b-', label='Message Rate')
    ax1_twin.plot(df['Timestamp'], df['AverageLatency'], 'r-', label='Avg Latency')
    
    ax1.set_xlabel('Time (s)')
    ax1.set_ylabel('Messages/Second', color='b')
    ax1_twin.set_ylabel('Latency (s)', color='r')
    
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax1_twin.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
    ax1.set_title('Message Rate and Latency over Time')
    
    # 2. Connection Status and Stability
    ax2 = axes[0, 1]
    ax2_twin = ax2.twinx()
    
    ax2.plot(df['Timestamp'], df['Connected'], 'g-', label='Connected')
    ax2_twin.plot(df['Timestamp'], df['Stability'], 'y-', label='Stability')
    
    ax2.set_xlabel('Time (s)')
    ax2.set_ylabel('Connection Status', color='g')
    ax2_twin.set_ylabel('Stability Score', color='y')
    
    lines1, labels1 = ax2.get_legend_handles_labels()
    lines2, labels2 = ax2_twin.get_legend_handles_labels()
    ax2.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
    ax2.set_title('Connection Status and Stability')
    
    # 3. Message Counts
    ax3 = axes[1, 0]
    ax3.plot(df['Timestamp'], df['MessagesSent'], label='Messages Sent')
    ax3.plot(df['Timestamp'], df['MessagesReceived'], label='Messages Received')
    ax3.set_xlabel('Time (s)')
    ax3.set_ylabel('Count')
    ax3.legend()
    ax3.set_title('Cumulative Message Counts')
    
    # 4. Packet Loss
    ax4 = axes[1, 1]
    ax4.plot(df['Timestamp'], df['PacketLoss'], 'r-')
    ax4.set_xlabel('Time (s)')
    ax4.set_ylabel('Packet Loss (%)')
    ax4.set_title('Packet Loss over Time')
    
    # 5. Time Since Last Communication
    ax5 = axes[2, 0]
    ax5.plot(df['Timestamp'], df['TimeSinceLastReceived'], label='Since Last Received')
    ax5.plot(df['Timestamp'], df['TimeSinceLastSent'], label='Since Last Sent')
    ax5.set_xlabel('Time (s)')
    ax5.set_ylabel('Time (s)')
    ax5.legend()
    ax5.set_title('Time Since Last Communication')
    
    # 6. System Stability Histogram
    ax6 = axes[2, 1]
    sns.histplot(data=df, x='Stability', bins=30, ax=ax6)
    ax6.set_xlabel('Stability Score')
    ax6.set_ylabel('Count')
    ax6.set_title('Stability Score Distribution')
    
    # Adjust layout and save
    plt.tight_layout()
    plt.savefig(f'{output_dir}/performance_dashboard.png', dpi=300, bbox_inches='tight')
    plt.close()

def generate_summary_stats(df):
    """Generate summary statistics from the performance data."""
    summary = {
        'Session Duration (s)': df['Timestamp'].max(),
        'Total Messages Sent': df['MessagesSent'].max(),
        'Total Messages Received': df['MessagesReceived'].max(),
        'Average Message Rate': df['MessageRate'].mean(),
        'Average Latency (s)': df['AverageLatency'].mean(),
        'Average Stability Score': df['Stability'].mean(),
        'Maximum Packet Loss': df['PacketLoss'].max(),
        'Connection Drops': len(df[df['Connected'] == 0])
    }
    
    return pd.Series(summary)

def main():
    """Main function to run the analysis."""
    # Load data using raw string for Windows path
    filepath = r"C:\Users\YuNakaura\Desktop\UnityHapticTouch3DTest\TestHaptic3Blocks\Assets\ExperimentData\system_performance__20241107_145125.csv"
    
    df = load_and_process_data(filepath)
    
    # Create plots
    create_performance_dashboard(df)
    
    # Generate and print summary statistics
    summary_stats = generate_summary_stats(df)
    print("\nSystem Performance Summary:")
    print("="*50)
    print(summary_stats.to_string())
    
    # Save summary statistics to file
    summary_stats.to_csv('plots/summary_statistics.csv')

if __name__ == "__main__":
    main()
```
